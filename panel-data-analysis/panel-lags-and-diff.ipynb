```python
import pandas as pd
```

```python
# Загрузим данные
df = pd.read_csv('./panels-plm/Wages.csv')
panel_df = df.set_index(['id', 'time'], drop=False)
panel_df.head(20)
```

```python
panel_df['d_wks'] = panel_df['wks'].groupby(level=0).diff()
panel_df.head(20)
```

```python
panel_df['lag_wks'] = panel_df['wks'].groupby(level=0).shift()
panel_df.head(20)
```
