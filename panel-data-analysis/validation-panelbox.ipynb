# Panel model validation

```python
import numpy as np
import pandas as pd

from panelbox import PooledOLS					# Pooled model
from panelbox import RandomEffects				# Random-effect model
from panelbox import FixedEffects				# Fixed-effect model

# heteroskedasticity tests
from panelbox.validation.heteroskedasticity import  ModifiedWaldTest, BreuschPaganTest, WhiteTest

# serial correlation tests
from panelbox.validation.serial_correlation import WooldridgeARTest, BreuschGodfreyTest, BaltagiWuTest

# cross-sectional correlation
from panelbox.validation.cross_sectional_dependence import PesaranCDTest, BreuschPaganLMTest, FreesTest

# panelbox validation (all tests)
from panelbox.validation import ValidationSuite


from panelbox.experiment import PanelExperiment	# Compare the results of multiple models
```

```python
# data importing
df = pd.read_csv('./panels-plm/Gasoline.csv')
```

```python
mod_fe = FixedEffects(formula='lgaspcar~lincomep+lrpmg+lcarpcap', data=df, entity_col='country', time_col='year')
res_fe = mod_fe.fit()

print(res_fe.summary())
```

## Heteroskedasticity tests

```python
# Modified Wald Test
mwald = ModifiedWaldTest(res_fe)
result_mwald = mwald.run(alpha=0.05)

print(result_mwald)
```

```python
# Breusch-Pagan Test
bp = BreuschPaganTest(res_fe)
result_bp = bp.run(alpha=0.05)
print(result_bp)
```

```python
# White Test
white = WhiteTest(res_fe)
result_white = white.run(alpha=0.05)
print(result_white)
```

## Serial Correlation Tests

```python
# Wooldridge AR Test
wooldridge = WooldridgeARTest(res_fe)
result_w = wooldridge.run(alpha=0.05)

print(result_w)
```

```python
# Breusch-Godfrey Test (lags 1 and 2)
bg = BreuschGodfreyTest(res_fe)

result_bg1 = bg.run(lags=1, alpha=0.05)
result_bg2 = bg.run(lags=2, alpha=0.05)

print("Breusch-Godfrey AR(1):")
print(result_bg1)

print("\nBreusch-Godfrey AR(2):")
print(result_bg2)
```

```python
# Baltagi-Wu LBI Test
bw = BaltagiWuTest(res_fe)
result_bw = bw.run(alpha=0.05)

print(result_bw)
```

## Cross-Sectional Dependence Tests

```python
# Pesaran CD Test
pesaran = PesaranCDTest(res_fe)
result_cd = pesaran.run(alpha=0.05)

print(result_cd)
```

```python
# Breusch-Pagan LM Test
bp_lm = BreuschPaganLMTest(res_fe)
result_bplm = bp_lm.run(alpha=0.05)
print(result_bplm)
```

```python
# Frees Test
frees = FreesTest(res_fe)
result_frees = frees.run(alpha=0.05)
print(result_frees)
```

## ValidationSuite: Unified Interface

```python
# Run ValidationSuite
suite = ValidationSuite(res_fe)
report = suite.run(tests="all", alpha=0.05)

print(report)
```
