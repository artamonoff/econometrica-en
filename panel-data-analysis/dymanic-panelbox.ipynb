# Panel data analysis with Python

```python
import numpy as np
import pandas as pd

from panelbox.models.dynamic import AndersonHsiao
from panelbox import DifferenceGMM

```

## Panel data importing

Consider `Gasoline` dataset. We load it from a file as a DataFrame

```python
df = pd.read_csv('./panels-plm/Gasoline.csv')
df.head()
```

## Anderson-Hsiao

[Link](https://panelbox.readthedocs.io/en/latest/api/dynamic-models/)

```python
# we use levels as instruments
mod_ah_lvls = AndersonHsiao(data=df, dep_var='lgaspcar', lags=1, exog_vars=['lincomep', 'lrpmg', 'lcarpcap'],
						entity_col='country', time_col='year', instrument='levels')

res_ah_lvls = mod_ah_lvls.fit()

# rho & betas
res_ah_lvls['rho'], res_ah_lvls['beta']
```

```python
# all coefficients
res_ah_lvls['params']
```

```python
# we use differences as instruments
mod_ah_diff= AndersonHsiao(data=df, dep_var='lgaspcar', lags=1, exog_vars=['lincomep', 'lrpmg', 'lcarpcap'],
						entity_col='country', time_col='year', instrument='levels')

res_ah_diff = mod_ah_diff.fit()

# rho & betas
res_ah_diff['rho'], res_ah_diff['beta']
```

## Arellano-Bond first-difference GMM

[Link](https://panelbox.readthedocs.io/en/latest/api/gmm/)

### Fitting

```python
mod_gmm = DifferenceGMM(data=df, dep_var='lgaspcar', lags=1, exog_vars=['lincomep', 'lrpmg', 'lcarpcap'], time_dummies=False,
						id_var='country', time_var='year', two_step=True, collapse=True)

res_gmm = mod_gmm.fit()
```

```python
print(res_gmm.summary())
```
