# Panel experiment

```python
import numpy as np
import pandas as pd

from panelbox.experiment import PanelExperiment	# Compare the results of multiple models
```

```python
# data importing
df = pd.read_csv('./panels-plm/Gasoline.csv')
df.head()
```

## Panel Experiment

We fit a collection of models and get a fitted report

```python
experiment = PanelExperiment(
	formula="lgaspcar~lincomep+lrpmg+lcarpcap",
	data=df,
	entity_col="country",
	time_col="year"
)

experiment.fit_model(model_type = 'pooled', name='PL', cov_type='clustered')
experiment.fit_model(model_type = 're', name='RE', cov_type='clustered')
experiment.fit_model(model_type = 'fe', name='FE', cov_type='clustered')

# experiment.fit_all_models(model_types=['pooled', 're', 'fe'], names=['PL', 'RE', 'FE'], cov_type="clustered")

comparison = experiment.compare_models()
print(comparison.summary())
```

```python
best = comparison.best_model("aic", prefer_lower=True)
print(f"Best model by AIC: {best}")
```

```python
val = experiment.validate_model(name = "FE")
print(val.summary())
```

```python
resid_analysis = experiment.analyze_residuals(name='FE')
print(resid_analysis.summary())
```
