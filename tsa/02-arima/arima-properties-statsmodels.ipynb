# ARMA model. Properties

Checking the stationarity and invertibility conditions for ARMA/ARIMA in the `statsmodels` package

We use the [`ArmaProcess`](https://www.statsmodels.org/stable/generated/statsmodels.tsa.arima_process.ArmaProcess.html#statsmodels.tsa.arima_process.ArmaProcess) class

```python
import numpy as np
from statsmodels.tsa.arima_process import ArmaProcess
```

Consider the ARMA(2,2) model

$$
	y_t=0.6 y_{t-1}-0.3 y_{t-2}+u_t+0.4u_{t-1}-0.3u_{t-2}
$$
Specification using the lag operator
$$
\begin{aligned}
	\Phi(L)y_t&=\Theta(L)u_t & \Phi(z)&=1-0.6z+0.3z^2 & \Theta(L)&=1+0.4z-0.3z^2
\end{aligned}
$$

Define the process via the coefficients of the polynomials $\Phi(z),\Theta(z)$, __starting with one__

```python
arima_process = ArmaProcess(ar = [1, -0.6, 0.3], ma=[1, 0.4, -0.3])
```

## Checking the stationarity condition

```python
arima_process.isstationary
```

roots of the characteristic polynomial $\Phi(z)$, including complex ones

```python
arima_process.arroots
```

roots of the characteristic polynomial $\Phi(z)$ in absolute value.

```python
np.abs(arima_process.arroots)
```

## Checking the invertibility condition

```python
arima_process.isinvertible
```

roots of the polynomial $\Theta(z)$, including complex ones

```python
arima_process.maroots
```

roots of the polynomial $\Theta(z)$ in absolute value

```python
np.abs(arima_process.maroots)
```
