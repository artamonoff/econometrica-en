# Unit root tests (the [`statsmodels`](https://www.statsmodels.org/stable/index.html) package)

```python
import numpy as np
import pandas as pd

from statsmodels.tsa.api import adfuller, kpss, range_unit_root_test

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

Load quarterly GDP data from the [`FRED`](https://fred.stlouisfed.org/) database (Symbol [`GDP`](https://fred.stlouisfed.org/series/GDP)) from 1990 Q1 to 2023 Q4 and create a series `y`=log(GDP)

```python
y = np.log( web.DataReader(name='GDP', data_source='fred', start='1990-1', end='2023-12') )
```

```python
ax = y.plot(title='US GDP')

# label on the x-axis
ax.set_xlabel('Date')
# label on the y-axis
ax.set_ylabel('log(GDP)')
# show the grid
ax.grid()
# remove the legend
ax.legend().remove()

plt.show()
```

## ADF test

Use the [`adfuller`](https://www.statsmodels.org/stable/generated/statsmodels.tsa.stattools.adfuller.html#statsmodels.tsa.stattools.adfuller) method

__Test parameters__

|Parameter|Value|Description|
|-|-|-|
|`maxlag`|`None, int`|Maximum number of lags (default $12*(T/100)^{1/4}$)|
|`regression`|`“c”,”ct”,”ctt”,”n”`|Test variant: _const, linear trend, quadratic trend, zero mean_|
|`autolag`|`“AIC”, “BIC”, “t-stat”, None`|Criterion for automatic lag selection|

Choose the test variant
* with trend
* with automatic lag selection by the BIC criterion

```python
adf_stat, pval, usedlag, nobs, critical_values, BIC = adfuller(y, regression='ct', autolag='BIC')
# test statistic, its p-value and critical values
adf_stat, pval, critical_values
```

## KPSS test

Use the [`kpss`](https://www.statsmodels.org/stable/generated/statsmodels.tsa.stattools.kpss.html#statsmodels.tsa.stattools.kpss) method

__Test parameters__

|Parameter|Value|Description|
|-|-|-|
|`regression`|`“c”,”ct”`|Test variant: _const, linear trend_|


Choose the test variant with trend

```python
kpss_stat, p_value, lags, crit = kpss(y, regression='ct')
# test statistic, its p-value and critical values
kpss_stat, p_value, crit
```

## RUR test

Use the [`range_unit_root_test`](https://www.statsmodels.org/stable/generated/statsmodels.tsa.stattools.range_unit_root_test.html#statsmodels.tsa.stattools.range_unit_root_test) method

__Important__: we test

$$
\begin{aligned}
	H_0&:I(0) & &vs & H_1&:I(1)
\end{aligned}
$$

(__no variant with trend!__)

Apply to the first difference

```python
rur_stat, p_value, crit = range_unit_root_test(y.diff().dropna())
# test statistic, its p-value and critical values
rur_stat, p_value, crit
```
