# Unit root tests (the [`arch`](https://arch.readthedocs.io/en/latest/index.html) package)

```python
import numpy as np
import pandas as pd

from arch.unitroot import ADF, KPSS, PhillipsPerron, DFGLS

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

Load weekly data from the [`FRED`](https://fred.stlouisfed.org/) database on Market Yield on U.S. Treasury Securities at 10-Year Constant Maturity (Symbol [`WGS10YR`](https://fred.stlouisfed.org/series/WGS10YR)) from 2000-01-01 to 2023-12-31 and create a dataframe `y`

```python
y = web.DataReader(name='WGS10YR', data_source='fred', start='2000-01-01', end='2023-12-31')
```

```python
ax = y.plot(title='Market Yield on U.S. Treasury Securities')

# label on the x-axis
ax.set_xlabel('Date')
# label on the y-axis
ax.set_ylabel('10Y Rate (%)')
# show the grid
ax.grid()
# remove the legend
ax.legend().remove()

plt.show()
```

## ADF test

Use the [`ADF`](https://arch.readthedocs.io/en/latest/unitroot/generated/arch.unitroot.ADF.html#arch.unitroot.ADF) method

Choose the test variant
* with a constant (without trend)
* with automatic lag selection by the BIC criterion

```python
ADF(y, trend='c', method='bic')
```

## KPSS test

Use the [`KPSS`](https://arch.readthedocs.io/en/latest/unitroot/generated/arch.unitroot.KPSS.html#arch.unitroot.KPSS) method

Choose the test variant with a constant (without trend)

```python
KPSS(y, trend='c')
```

## PP test

Use the [`PhillipsPerron`](https://arch.readthedocs.io/en/latest/unitroot/generated/arch.unitroot.PhillipsPerron.html#arch.unitroot.PhillipsPerron) method

Choose the test variant
* with a constant (without trend)
* tau test statistic

```python
PhillipsPerron(y, trend='c', test_type='tau')
```

## DF-GLS test

Use the [`DFGLS`](https://arch.readthedocs.io/en/latest/unitroot/generated/arch.unitroot.DFGLS.html#arch.unitroot.DFGLS) method

Choose the test variant
* with a constant (without trend)
* with automatic lag selection by the BIC criterion

```python
DFGLS(y, trend='c', method='bic')
```
