# ARIMA model: automatic hyperparameter tuning (the [`sktime`](https://www.sktime.net/en/stable/) package)

```python
import numpy as np
import pandas as pd

# import from the pmdarima package
from sktime.forecasting.arima import AutoARIMA

from sktime.utils.plotting import plot_series
# forecasting horizon
from sktime.forecasting.base import ForecastingHorizon

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

Load weekly data from the [`FRED`](https://fred.stlouisfed.org/) database on the 3-Month Treasury Bill Secondary Market Rate (Symbol [`WTB3MS`](https://fred.stlouisfed.org/series/WTB3MS)) from 2000-01-01 to 2023-12-31 and create a dataframe `y`

```python
y = web.DataReader(name='WTB3MS', data_source='fred', start='2000-01-01', end='2023-12-31')
```

# Automatic selection of the model order

Basic parameters of automatic selection

|Criterion|`AutoARIMA`|Values|
|-|-|-|
|Information criterion|`information_criterion`|aic (default), aicc, bic, hqic |
|unit root test|`test`|kpss (default), adf, pp|
|max order d|`max_d`| 2 by default |
|max order p,q|`max_p`, ` max_q`|5 by default|
|seasonal model|`seasonal`|True by default|
|significance level|`alpha`|0.05 by default|

__Note__: whether to include drift/constant/trend in the model is selected automatically

```python
forecaster = AutoARIMA(information_criterion='bic', test='kpss', seasonal=False)
forecaster.fit(y)
forecaster.get_fitted_params() #['order']
```
