# Unit root tests (the [`sktime`](https://www.sktime.net/en/stable/index.html) package)

```python
import numpy as np
import pandas as pd

from sktime.param_est.stationarity import StationarityADF, StationarityKPSS, StationarityDFGLS, \
	StationarityPhillipsPerron, StationarityADFArch, StationarityKPSSArch

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show InterpolationWarning from statsmodels
from statsmodels.tools.sm_exceptions import InterpolationWarning
warnings.simplefilter('ignore', category=InterpolationWarning)
```

Load monthly M2 data from the [`FRED`](https://fred.stlouisfed.org/) database (Symbol [`M2SL`](https://fred.stlouisfed.org/series/M2SL)) from 2000-01 to 2023-12 and create a dataframe `y`=log(M2)

```python
y = np.log( web.DataReader(name='M2SL', data_source='fred', start='2000-01', end='2023-12') )
```

```python
ax = y.plot(title='M2 Monthly data')

# label on the x-axis
ax.set_xlabel('Date')
# label on the y-axis
ax.set_ylabel('log(M2)')
# show the grid
ax.grid()
# remove the legend
ax.legend().remove()

plt.show()
```

## ADF test

### Variant 1

Use the [`StationarityADF`](https://www.sktime.net/en/stable/api_reference/auto_generated/sktime.param_est.stationarity.StationarityADF.html#sktime.param_est.stationarity.StationarityADF) method (actually uses the method from `statsmodels`)

Choose the test variant
* with trend
* with automatic lag selection by the BIC criterion

```python
sty_est = StationarityADF(p_threshold=0.05, regression='ct', autolag='BIC')
sty_est.fit(y) 
sty_est.get_fitted_params()["stationary"]  
```

```python
# full test report
sty_est.get_fitted_params()
```

### Variant 2

Use the [`StationarityADFArch`](https://www.sktime.net/en/stable/api_reference/auto_generated/sktime.param_est.stationarity.StationarityADFArch.html#sktime.param_est.stationarity.StationarityADFArch) method (actually uses the method from `arch`)

Choose the test variant
* with trend
* with automatic lag selection by the BIC criterion

```python
sty_est = StationarityADFArch(trend='ct', method='bic', p_threshold=0.05)
sty_est.fit(y) 
sty_est.get_fitted_params()["stationary"]
```

```python
# full test report
sty_est.get_fitted_params()
```
