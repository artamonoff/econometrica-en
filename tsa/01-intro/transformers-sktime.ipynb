# Transformers (the [`sktime`](https://www.sktime.net/en/stable/) package)

[List of transformers](https://www.sktime.net/en/stable/api_reference/transformations.html) in sktime

```python
import numpy as np
import pandas as pd

# Basic transformers
from sktime.transformations.series.boxcox import LogTransformer, BoxCoxTransformer
from sktime.transformations.series.difference import Differencer
# creating a pipeline
from sktime.pipeline import make_pipeline

from sktime.utils.plotting import plot_series

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

```python
y = web.DataReader(name='M2SL', data_source='fred', start='1990-01')
```

```python
# log transformation
transformer = LogTransformer()
y_log = transformer.fit_transform(y)
y_log
```

```python
# differencing
transformer = Differencer(lags=1, na_handling='drop_na')
y_diff = transformer.fit_transform(y_log)
y_diff
```

```python
# create a pipeline from several consecutive transformations: log return
pipe = make_pipeline(LogTransformer(), Differencer(lags=1, na_handling='drop_na'))
y_pipe = pipe.fit_transform(y)
y_pipe
```

```python
# alternative way to create a pipeline
pipe = LogTransformer() * Differencer(lags=1, na_handling='drop_na')
y_pipe = pipe.fit_transform(y)
y_pipe
```

```python
# pipeline with Box-Cox transformation
pipe = BoxCoxTransformer() * Differencer(lags=1, na_handling='drop_na')
y_pipe = pipe.fit_transform(y)
y_pipe
```
