# Loading data from the FRED database

Importing libraries

```python
import numpy as np
import pandas as pd 

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

## Example 1 GDP
Load quarterly GDP data from the [`FRED`](https://fred.stlouisfed.org/) database (Symbol [`GDP`](https://fred.stlouisfed.org/series/GDP)) from 1980 Q1 to 2023 Q4 and create a dataframe `gdp`

```python
gdp = web.DataReader(name='GDP', data_source='fred', start='1980-1', end='2023-12' )
gdp
```

change the index to a (period) quarterly one

```python
gdp.index = gdp.index.to_period(freq='Q')
gdp
```

series plot

```python
gdp.plot()
```

change the plot settings

```python
ax = gdp.plot(title='GDP')
# label on the x-axis
ax.set_xlabel('Date')
# label on the y-axis
ax.set_ylabel('US GDP')
# show the grid
ax.grid()
# remove the legend
ax.legend().remove()

plt.show()
```

## Example 2: rates
Load monthly data from the [`FRED`](https://fred.stlouisfed.org/) database on 

* 3-Month Treasury Bill Secondary Market Rate (Symbol [`TB3MS`](https://fred.stlouisfed.org/series/TB3MS)) 
* 1-Year Treasury Bill Secondary Market Rate (Symbol [`TB1YR`](https://fred.stlouisfed.org/series/TB1YR)) 
* Market Yield on U.S. Treasury Securities at 10-Year Constant Maturity (Symbol [`GS10`](https://fred.stlouisfed.org/series/GS10))

from 1990-01 to 2023-12 and create a dataframe `rates`

```python
rates = web.DataReader(name=['TB3MS', 'TB1YR', 'GS10'], data_source='fred',start='1990-01', end='2023-12')
rates
```

set a (period) monthly index and rename the variables

```python
# override the index
rates.index = rates.index.to_period(freq='M')
# rename the variables/columns
rates.columns = ['3M_rate', '1Y_rate', '10Y_rate']
rates
```

Visualization (all on a single plot)

```python
ax = rates.plot()

ax.set_xlabel('Date')
ax.set_ylabel('Rate (%)')
ax.grid()

plt.show()
```

visualize as separate plots for each variable

```python
rates.plot(subplots=True)
```

change the plot settings

```python
axs = rates.plot(subplots=True)

axs[0].set_xlabel('Date')
axs[0].set_ylabel('3M rate')
axs[0].grid()
axs[0].legend().remove()

axs[1].set_xlabel('Date')
axs[1].set_ylabel('1Y rate')
axs[1].grid()
axs[1].legend().remove()

axs[2].set_xlabel('Date')
axs[2].set_ylabel('10Y rate')
axs[2].grid()
axs[2].legend().remove()

plt.show()
```
