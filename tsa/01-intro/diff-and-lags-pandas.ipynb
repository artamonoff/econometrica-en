# Differences and lags

```python
import numpy as np
import pandas as pd 

import pandas_datareader.data as web

from sktime.utils.plotting import plot_series, plot_lags

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

## Example 1 (first difference)
Load monthly M2 data from the [`FRED`](https://fred.stlouisfed.org/) database (Symbol [`M2SL`](https://fred.stlouisfed.org/series/M2SL)) from 1990-01 to 2023-12 and create a dataframe `m2`

```python
y = web.DataReader(name='M2SL', data_source='fred', start='1990-01', end='2023-12')
y.index = y.index.to_period(freq='M')
y.columns = ['M2']
y
```

```python
# first difference
y.diff()
```

```python
# second difference
y.diff().diff()
```

```python
# log return
np.log(y).diff()
```

Visualize with different styles (see predefined styles [here](https://matplotlib.org/stable/gallery/style_sheets/style_sheets_reference.html))

```python
plt.style.use('ggplot')

ax = y.plot()
ax.set_title('US M2 (monthly data)')
ax.set_ylabel('M2')
ax.set_xlabel('Date')
ax.legend().remove()

plt.show()
```

```python
plt.style.use('grayscale')

ax = y.diff().plot()
ax.set_title('First difference of M2 (monthly data)')
ax.set_ylabel('diff((M2)')
ax.legend().remove()

plt.show()
```

```python
plt.style.use('bmh')

ax = np.log(y).diff().plot()
ax.set_title('Log return of M2 (monthly data)')
ax.set_ylabel('log(diff((M2))')
ax.legend().remove()

plt.show()
```

```python
# histogram for log return
plt.style.use('default')
ax = np.log(y).diff().plot.hist(bins=30)
ax.set_title('Histogram for log return of M2')
ax.legend().remove()

plt.show()
```

## Example 2 (lags)

Load monthly data from the [`FRED`](https://fred.stlouisfed.org/) database on the 3-Month Treasury Bill Secondary Market Rate (Symbol [`TB3MS`](https://fred.stlouisfed.org/series/TB3MS)) from 1990-01 to 2023-12

```python
y = web.DataReader(name='TB3MS', data_source='fred',start='1990-01', end='2023-12')
y.index = y.index.to_period(freq='M')
y.columns = ['rate']
y
```

```python
plot_series(y)
```

```python
plot_lags(y, lags=1)

plt.show()
```

```python
plot_lags(y.diff(), lags=1)

plt.show()
```

```python
plot_lags(y, lags=[1,2])

plt.show()
```

```python
plot_lags(y.diff(), lags=[1,2])

plt.show()
```

```python
# Add a lag to the dataset
y['lag_rate'] = y['3Mrate'].shift()
y.corr()
```
