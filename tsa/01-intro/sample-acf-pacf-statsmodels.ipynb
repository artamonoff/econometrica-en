# Sample ACF and PACF (the [`statsmodels`](https://www.statsmodels.org/stable/index.html) package)

```python
import numpy as np
import pandas as pd

from statsmodels.tsa.api import acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

Load weekly data from the [`FRED`](https://fred.stlouisfed.org/) database on Market Yield on U.S. Treasury Securities at 10-Year Constant Maturity (Symbol [`WGS10YR`](https://fred.stlouisfed.org/series/WGS10YR)) from 2000-01-01 to 2023-12-31 and create a dataframe `y`

```python
y = web.DataReader(name='WGS10YR', data_source='fred', start='2000-01-01', end='2023-12-31')
```

## Sample ACF & PACF

compute (with rounding) up to lag 7 

```python
acf(y, nlags=7, missing = 'drop').round(3)
```

```python
pacf(y, nlags=7).round(3)
```

ACF visualization

```python
plot_acf(y, lags=35, alpha=0.05, missing = 'drop', zero=True, title='Autocorrelation for levels')

plt.show()
```

PACF visualization

```python
plot_pacf(y, lags=35, alpha=0.05, zero=True, title='Partial Autocorrelation for levels')

plt.show()
```

Combine the two plots into one

```python
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

plot_acf(y, ax=axs[0], lags=30, alpha=0.05, missing = 'drop', zero=True, title='Autocorrelation for levels')

plot_pacf(y, ax=axs[1], lags=30, alpha=0.05, zero=True, title='Partial Autocorrelation for levels')

plt.show()
```

## Using ACF and PACF to check series stationarity

For a stationary series, the ACF & PACF coefficients should decay "quickly" to zero

Compare the behavior of ACF & PACF for
* series levels
* first difference

Create the series and its first differences and remove missing observations

```python
dy = y.diff().dropna()
```

```python
fig, axs = plt.subplots(nrows=2, ncols=1)

y.plot(ax=axs[0])
dy.plot(ax=axs[1])

axs[0].legend().remove()
axs[1].legend().remove()

axs[0].set_ylabel('levels')
axs[1].set_ylabel('diff')

plt.show()
```

Plot ACF & PACF for each series and combine into one figure

```python
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))

plot_acf(y, ax=axs[0,0], lags=30, alpha=0.05, missing = 'drop', zero=True, title='Autocorrelation for levels')
plot_pacf(y, ax=axs[0, 1], lags=30, alpha=0.05, zero=True, title='Partial Autocorrelation for levels')

plot_acf(dy, ax=axs[1,0], lags=30, alpha=0.05, missing = 'drop', zero=True, title='Autocorrelation for diff')
plot_pacf(dy, ax=axs[1, 1], lags=30, alpha=0.05, zero=True, title='Partial Autocorrelation for diff')

plt.show()
```
