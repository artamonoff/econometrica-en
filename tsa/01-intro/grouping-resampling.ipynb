# Grouping and aggregating data

```python
import numpy as np
import pandas as pd 

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

## Example 1 Short-term rate
Load weekly data from the [`FRED`](https://fred.stlouisfed.org/) database on the 3-Month Treasury Bill Secondary Market Rate (Symbol [`WTB3MS`](https://fred.stlouisfed.org/series/WTB3MS)) from 2000-01-01 to 2023-12-31 and create a dataframe `rate3m_weekly`

```python
y_weekly = web.DataReader(name='WTB3MS', data_source='fred', start='2000-01-01', end='2023-12-31')
y_weekly
```

Convert (aggregate) to quarterly data by averaging

```python
y_quarterly = y_weekly.groupby(pd.Grouper(freq='Q')).mean()
y_quarterly.index = y_quarterly.index.to_period(freq='Q')
y_quarterly
```

## Example 2 Long-term rate
Load daily data from the [`FRED`](https://fred.stlouisfed.org/) database on the Market Yield on U.S. Treasury Securities at 10-Year Constant Maturity (Symbol [`DGS10`](https://fred.stlouisfed.org/series/DGS10)) from 2000-01-01 to 2023-12-31 and create a dataframe `rate10_daily`

```python
y_daily = web.DataReader(name='DGS10', data_source='fred', start='2000-01-01', end='2023-12-31')
y_daily
```

Convert (aggregate) to monthly data by taking the last observation of the month

```python
y_monthly = y_daily.groupby(pd.Grouper(freq='M')).tail(1)
y_monthly.index = y_monthly.index.to_period(freq='M')
y_monthly
```
