# Sample ACF and PACF (the [`sktime`](https://www.sktime.net/en/stable/get_started.html) package)

```python
import numpy as np
import pandas as pd

# Sample ACF & PACF as transformers
from sktime.transformations.series.acf import AutoCorrelationTransformer, PartialAutoCorrelationTransformer
# Basic transformers
from sktime.transformations.series.boxcox import LogTransformer, BoxCoxTransformer
from sktime.transformations.series.difference import Differencer
# Visualization
from sktime.utils.plotting import plot_correlations

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

Load weekly data from the [`FRED`](https://fred.stlouisfed.org/) database on Market Yield on U.S. Treasury Securities at 10-Year Constant Maturity (Symbol [`WGS10YR`](https://fred.stlouisfed.org/series/WGS10YR)) from 2000-01-01 to 2023-12-31 and create a dataframe `y`

```python
y = web.DataReader(name='WGS10YR', data_source='fred', start='2000-01-01', end='2023-12-31')
```

## Visualization of sample ACF & PACF

Visualize the series itself and the sample ACF & PACF for the original series

```python
plot_correlations(y, lags=25, zero_lag=True, alpha=0.05, suptitle='Market Yield', acf_title='ACF', pacf_title='PACF')

plt.show()
```

Compute the first 5 ACF coefficients

```python
acf_transformer = AutoCorrelationTransformer(n_lags=5)
acf_transformer.fit_transform(y)
```

Compute the first 5 PACF coefficients

```python
pacf_transformer = PartialAutoCorrelationTransformer(n_lags=5)
pacf_transformer.fit_transform(y)
```

## Sample ACF & PACF as parts of a pipeline

```python
z = web.DataReader(name='GDP', data_source='fred', start='1990-01-01')
```

```python
pipe_acf = LogTransformer() * Differencer(lags=1, na_handling='drop_na') * AutoCorrelationTransformer(n_lags=5)
pipe_acf.fit_transform(z)
```

```python
pipe_pacf = LogTransformer() * Differencer(lags=1, na_handling='drop_na') * PartialAutoCorrelationTransformer(n_lags=5)
pipe_pacf.fit_transform(z)
```

```python
pipe = LogTransformer() * Differencer(lags=1, na_handling='drop_na')
z_pipe = pipe.fit_transform(z)

plot_correlations(z_pipe, lags=25, zero_lag=True, alpha=0.05, suptitle='GDP', acf_title='ACF', pacf_title='PACF')

plt.show()
```
