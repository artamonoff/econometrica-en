# SARIMA Model (package [`statsmodels`](https://www.statsmodels.org/stable/index.html))

```python
import numpy as np
import pandas as pd

from statsmodels.tsa.api import ARIMA
# alternative
# from statsmodels.tsa.api import SARIMAX
from statsmodels.graphics.tsaplots import plot_predict

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

Load daily data from the [`FRED`](https://fred.stlouisfed.org/) database on Gross Domestic Product () (Symbol [`NA000334Q`](https://fred.stlouisfed.org/series/NA000334Q)) from 1990-01-01 to 2025-12-31 (series `z`) and create dataframe `y=log(z)`

```python
z = web.DataReader(name='NA000334Q', data_source='fred', start='1990-01-01', end='2025-12-31')
y = np.log(z)

y.plot()
plt.show()
```

## Fitting SARIMA of a given order

Fit a SARMA(1,0,1)(1,1,1,4) model with linear trend for `y` (set the number of seasons to 4)

Specification

$$
	(1-\phi L)(1-\Phi L^4)(1-L^4) (y_t-\gamma t)=(1+\theta L)(1+\Theta L^4)u_t
$$

```python
# model specification
mod = ARIMA(y, order=(1,0,1), seasonal_order=(1,1,1,4), trend='t', missing='drop')
# fit the model to the data
res = mod.fit()
# output fitting results
res.summary(alpha=0.05)
```

## Forecasting

Build a forecast using the SARMA(1,1,1)(1,0,1,4) model with linear trend for `y` (set the number of seasons to 4)

Specification

$$
	(1-\phi L)(1-\Phi L^4)(1-L) (y_t-\gamma t)=(1+\theta L)(1+\Theta L^4)u_t
$$

```python
# model specification
mod = ARIMA(y, order=(1,1,1), seasonal_order=(1,0,1,4), trend='t', missing='drop')
# fit the model to the data
res = mod.fit()
# forecasting
forecasts = res.forecast(steps=10)
forecasts
```

forecast visualization

```python
plot_predict(res, start=len(y), end=len(y)+10, alpha=0.05)

plt.show()
```

```python
plt.plot(y.tail(30), label='y')
plt.plot(forecasts, label='y_pred')
plt.legend()

plt.show()
```
