# Forecasting using STL decomposition into trend and seasonality (package [`sktime`](https://www.sktime.net/en/stable/))

```python
import numpy as np
import pandas as pd

from sktime.forecasting.trend import STLForecaster
# model for forecasting the series component
from sktime.forecasting.arima import StatsModelsARIMA as ARIMA

from sktime.utils.plotting import plot_series

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

Load daily data from the [`FRED`](https://fred.stlouisfed.org/) database on Gross Domestic Product () (Symbol [`NA000334Q`](https://fred.stlouisfed.org/series/NA000334Q)) from 1990-01-01 to 2025-12-31 (series `z`) and create dataframe `y=log(z)`

```python
z = web.DataReader(name='NA000334Q', data_source='fred', start='1990-01-01', end='2025-12-31')
z.index = z.index.to_period(freq='Q')
y = np.log(z)
```

```python
forecaster = STLForecaster(sp=4, forecaster_trend=ARIMA(order=(1,1,1), trend='t'), 
							forecaster_seasonal=ARIMA(order=(1,0,1), seasonal_order=(1,0,1,4), trend='n'),
							forecaster_resid=None)
y_pred = forecaster.fit_predict(y=y, fh=np.arange(1,11))
```

```python
plot_series(y.tail(50), y_pred)
plt.show()
```
