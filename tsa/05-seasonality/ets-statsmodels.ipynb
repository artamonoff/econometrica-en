# ETS Model (package [`statsmodels`](https://www.statsmodels.org/stable/index.html))

```python
import numpy as np
import pandas as pd

from statsmodels.tsa.api import ETSModel

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

Load daily data from the [`FRED`](https://fred.stlouisfed.org/) database on Gross Domestic Product () (Symbol [`NA000334Q`](https://fred.stlouisfed.org/series/NA000334Q)) from 1990-01-01 to 2025-12-31 (series `z`) and create dataframe `y=log(z)`

```python
z = web.DataReader(name='NA000334Q', data_source='fred', start='1990-01-01', end='2025-12-31')
y = np.log(z)

y.plot()
plt.show()
```

### Additive model for `y`

Fit an ETS(A,A,A) model and build a forecast. Set the number of seasons to 4

```python
# model specification
mod = ETSModel(y.squeeze(), error='add', trend='add', seasonal='add', seasonal_periods=4, damped_trend=False)
# fit the model to the data
res = mod.fit()
# output fitting results
res.summary(alpha=0.05)
```

```python
# forecasting
forecasts = res.forecast(steps=10)
forecasts
```

Forecast visualization

```python
plt.plot(y.tail(30), label='y')
plt.plot(forecasts, label='y_pred')
plt.legend()
plt.show()
```

Fit an ETS(M,M,M) model and build a forecast. Set the number of seasons to 4
