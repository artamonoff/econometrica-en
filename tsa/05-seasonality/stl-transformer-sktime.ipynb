# STL decomposition into trend and seasonality 

```python
import numpy as np
import pandas as pd

from sktime.transformations.series.detrend import STLTransformer

from sktime.utils.plotting import plot_series

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

Load daily data from the [`FRED`](https://fred.stlouisfed.org/) database on Gross Domestic Product (Symbol [`NA000334Q`](https://fred.stlouisfed.org/series/NA000334Q)) from 1990-01-01 to 2025-12-31 (series `z`) and create dataframe `y=log(z)`

```python
z = web.DataReader(name='NA000334Q', data_source='fred', start='1990-01-01',  end='2025-12-31')
y = np.log(z)
```

## Trend extraction

```python
transformer = STLTransformer(trend_deg=1, sp=4, return_components=False)
y_trend = transformer.fit_transform(y)

plot_series(y, y_trend, labels=['y', 'trend'])
plt.show()
```

## Decomposing the series into components

__Note__: in the final decomposition `transformed=trend+resid`

```python
transformer = STLTransformer(sp=4, return_components=True)
decomposition = transformer.fit_transform(y)
decomposition
```

```python
decomposition.plot(subplots=True)
plt.show()
```
