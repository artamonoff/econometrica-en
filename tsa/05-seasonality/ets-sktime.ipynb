# ETS Model (package [`sktime`](https://www.sktime.net/en/stable/))

```python
import numpy as np
import pandas as pd

from sktime.forecasting.ets import AutoETS

from sktime.utils.plotting import plot_series

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

Load daily data from the [`FRED`](https://fred.stlouisfed.org/) database on Gross Domestic Product () (Symbol [`NA000334Q`](https://fred.stlouisfed.org/series/NA000334Q)) from 1990-01-01 to 2025-12-31 (series `z`) and create dataframe `y=log(z)`

```python
z = web.DataReader(name='NA000334Q', data_source='fred', start='1990-01-01', end='2025-12-31')
z.index = z.index.to_period(freq='Q')
y = np.log(z)

plot_series(y)
plt.show()
```

## Fitting an ETS model with a given specification

Fit an ETS(A,A,A) model for `y` and build a forecast. Set the number of seasons to 4

```python
forecaster = AutoETS(error='add', trend='add', seasonal='add', damped_trend=False, auto=False, sp=4)
y_pred = forecaster.fit_predict(y=y, fh=np.arange(1, 11))
y_pred
```

```python
plot_series(y.tail(50), y_pred, labels=['y', 'y_pred'])
plt.show()
```

## Selecting the optimal ETS model

Select the optimal ETS model for `y`. Set the number of seasons to 4

```python
forecaster = AutoETS(auto=True, sp=4, information_criterion='aic', allow_multiplicative_trend=True)
forecaster.fit(y)
forecaster.summary()
```

Select the optimal ETS model for `z`. Set the number of seasons to 4

```python
forecaster = AutoETS(auto=True, sp=4, information_criterion='bic', allow_multiplicative_trend=True)
forecaster.fit(z)
forecaster.summary()
```
