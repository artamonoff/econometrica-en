```python
import numpy as np
import pandas as pd

from statsmodels.tsa.seasonal import STL

import pandas_datareader.data as web

# visualization settings
import matplotlib.pyplot as plt

# Do not show Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
# Do not show ValueWarning, ConvergenceWarning from statsmodels
from statsmodels.tools.sm_exceptions import ValueWarning, ConvergenceWarning
warnings.simplefilter('ignore', category=ValueWarning)
warnings.simplefilter('ignore', category=ConvergenceWarning)
```

Load daily data from the [`FRED`](https://fred.stlouisfed.org/) database on Gross Domestic Product () (Symbol [`NA000334Q`](https://fred.stlouisfed.org/series/NA000334Q)) from 1990-01-01 to 2025-12-31 (series `z`) and create dataframe `y=log(z)`

```python
z = web.DataReader(name='NA000334Q', data_source='fred', start='1990-01-01', end='2025-12-31')
y = np.log(z)
```

```python
stl = STL(y, seasonal=5)
res = stl.fit()
```

```python
res.plot()
plt.show()
```
